## 3. Clean the Dataset

Following the Day 10 workflow notes, cleaning includes missing values, incorrect data types, and invalid data. The notes specifically demonstrate converting dates with `pd.to_datetime()` and removing/filling missing values. fileciteturn3file0L2-L8

Here, `Flight_Date` is converted to datetime and the seven missing `Delay_Minutes` values are filled with the median delay so that delay-based analysis can use all 180 flights.

```python
df["Flight_Date"] = pd.to_datetime(df["Flight_Date"])

missing_delay_before = df["Delay_Minutes"].isna().sum()
delay_median = df["Delay_Minutes"].median()

df["Delay_Minutes"] = df["Delay_Minutes"].fillna(delay_median)

print("Missing Delay_Minutes before cleaning:", missing_delay_before)
print("Median delay used for filling:", delay_median)
print("Missing Delay_Minutes after cleaning:", df["Delay_Minutes"].isna().sum())
```

## 4. Transform the Data

```python
# Create useful derived columns.
df["Estimated_Revenue"] = df["Passengers"] * df["Average_Ticket_Price"]
df["Load_Factor_Pct"] = (df["Passengers"] / df["Seat_Capacity"]) * 100
df["Flight_Month"] = df["Flight_Date"].dt.month
df["Flight_Month_Name"] = df["Flight_Date"].dt.month_name()
df["Day_of_Week"] = df["Flight_Date"].dt.day_name()

display(df[
    ["Flight_ID", "Flight_Date", "Passengers", "Seat_Capacity",
     "Average_Ticket_Price", "Estimated_Revenue", "Load_Factor_Pct"]
].head(10))
```

## 5. Dataset-Level Summary
