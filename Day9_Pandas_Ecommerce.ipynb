## 1. Inspect the Datasets

```python
print("Customers")
display(customers.head())

print("Products")
display(products.head())

print("Orders")
display(orders.head())
```

## 2. Using `concat()`

The Orders dataset is split into two portions and recombined using `pd.concat()`.


```python
orders_part1 = orders.iloc[:60].copy()
orders_part2 = orders.iloc[60:].copy()

orders_combined = pd.concat([orders_part1, orders_part2], ignore_index=True)

print("First part:", orders_part1.shape)
print("Second part:", orders_part2.shape)
print("Combined:", orders_combined.shape)
display(orders_combined.head())
```

## 3. Using `merge()`

Customer and product information is joined to the Orders dataset using `Customer_ID` and `Product_ID`.


```python
processed = orders_combined.merge(
    customers,
    on="Customer_ID",
    how="left"
).merge(
    products,
    on="Product_ID",
    how="left"
)

display(processed.head())
print("Processed shape:", processed.shape)
```

## 4. Date and Time Operations

The order date is converted to datetime format and useful date components are extracted.


```python
processed["Order_Date"] = pd.to_datetime(processed["Order_Date"])

processed["Order_Month"] = processed["Order_Date"].dt.month
processed["Order_Month_Name"] = processed["Order_Date"].dt.month_name()
processed["Order_Day"] = processed["Order_Date"].dt.day
processed["Day_of_Week"] = processed["Order_Date"].dt.day_name()

display(processed[
    ["Order_Date", "Order_Month", "Order_Month_Name",
     "Order_Day", "Day_of_Week"]
].head(10))
```

## 5. Using `apply()`

`apply()` is used to calculate total sales and classify orders by value.


```python
processed["Total_Sales"] = processed.apply(
    lambda row: row["Quantity"] * row["Unit_Price"],
    axis=1
)

processed["Order_Value_Category"] = processed["Total_Sales"].apply(
    lambda x: "High" if x >= 5000 else ("Medium" if x >= 2000 else "Low")
)

display(processed[
    ["Order_ID", "Quantity", "Unit_Price",
     "Total_Sales", "Order_Value_Category"]
].head(10))
```

## 6. Basic Analysis

```python
print("Total sales:", processed["Total_Sales"].sum())
print("Average order value:", round(processed["Total_Sales"].mean(), 2))

print("\nSales by category:")
display(
    processed.groupby("Category")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
)

print("\nOrders by membership type:")
display(processed["Membership_Type"].value_counts())
```

## 7. Final Processed Dataset

```python
column_order = [
    "Order_ID", "Order_Date", "Order_Month", "Order_Month_Name",
    "Order_Day", "Day_of_Week", "Customer_ID", "Customer_Name",
    "City", "Region", "Membership_Type", "Product_ID", "Product_Name",
    "Category", "Brand", "Unit_Price", "Quantity", "Total_Sales",
    "Order_Value_Category", "Payment_Method", "Order_Status"
]

processed = processed[column_order]

display(processed.head(10))
print("Rows:", len(processed))
print("Columns:", len(processed.columns))
```

## 8. Export the Final CSV

Running this cell creates the final processed CSV file in the Colab session.


```python
output_file = "Day9_Final_Processed_Ecommerce_Dataset.csv"
processed.to_csv(output_file, index=False)

print(f"Saved: {output_file}")
```
