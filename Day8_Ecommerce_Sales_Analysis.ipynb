# Day 8 – E-commerce Sales Analysis

## Objective
Analyze the provided e-commerce sales dataset using Python and Pandas.

This notebook demonstrates:
- DataFrame creation and inspection
- Data selection using `.loc` and `.iloc`
- Filtering with conditions
- Sorting with `sort_values()`
- Grouping with `groupby()`
- Aggregation using `sum()`, `mean()`, `max()`, `min()`, and `count()`
- Analysis by category, city, product, and payment method
- Meaningful observations from the results


```python
import pandas as pd
import os

file_name = "Day8_Ecommerce_Sales_Dataset.csv"

# If the CSV is not already in the Colab working folder, upload it.
if not os.path.exists(file_name):
    from google.colab import files
    uploaded = files.upload()
    file_name = next(iter(uploaded))

df = pd.read_csv(file_name)
print("Dataset loaded successfully.")
print("Shape:", df.shape)
```

## 1. Inspect the Dataset

```python
print("First 5 rows:")
display(df.head())

print("Dataset information:")
df.info()

print("\nColumn names:")
print(df.columns.tolist())

print("\nMissing values:")
print(df.isnull().sum())
```

## 2. Data Selection

The course notes cover selecting columns and rows with `.loc` and `.iloc`.


```python
# Select one column
display(df["Product"].head())

# Select multiple columns
display(df[["Order_ID", "Customer", "Product", "Total_Sales"]].head())

# Select rows and columns using .loc
display(df.loc[0:4, ["Order_ID", "Product", "Total_Sales"]])

# Select rows and columns using .iloc
display(df.iloc[0:5, 0:5])
```

## 3. Filtering the Data

```python
# Orders with sales greater than 5000
sales_above_5000 = df.loc[df["Total_Sales"] > 5000]
display(sales_above_5000[["Order_ID", "Product", "Total_Sales"]])

# Orders with quantity of 3 or more
quantity_3_plus = df.loc[df["Quantity"] >= 3]
display(quantity_3_plus[["Order_ID", "Product", "Quantity"]].head())

# Orders from the Sports category
sports_orders = df.loc[df["Category"] == "Sports"]
display(sports_orders[["Order_ID", "Product", "Category", "Total_Sales"]].head())

# Multiple conditions
sports_high_sales = df.loc[
    (df["Category"] == "Sports") & (df["Total_Sales"] > 5000)
]
display(sports_high_sales[["Order_ID", "Product", "Total_Sales"]])

# Orders paid by UPI or Credit Card
selected_payments = df.loc[
    df["Payment_Method"].isin(["UPI", "Credit Card"])
]
display(selected_payments[["Order_ID", "Payment_Method", "Total_Sales"]].head())
```

## 4. Date and Value Filtering

```python
# Convert Order_Date to datetime
df["Order_Date"] = pd.to_datetime(df["Order_Date"])

# Orders placed between two dates
date_filtered = df.loc[
    df["Order_Date"].between("2026-02-01", "2026-03-31")
]
display(date_filtered[["Order_ID", "Order_Date", "Product", "Total_Sales"]].head())

# Sales between 2000 and 8000
sales_range = df.loc[df["Total_Sales"].between(2000, 8000)]
display(sales_range[["Order_ID", "Product", "Total_Sales"]].head())
```

## 5. String Filtering

```python
# Products containing the word 'Wireless'
wireless_products = df.loc[
    df["Product"].str.contains("Wireless", case=False, na=False)
]
display(wireless_products[["Order_ID", "Product", "Total_Sales"]])

# Customers whose names start with 'A'
customers_a = df.loc[
    df["Customer"].str.startswith("A", na=False)
]
display(customers_a[["Customer", "Product", "Total_Sales"]].head())

# Products whose names end with 'Set'
set_products = df.loc[
    df["Product"].str.endswith("Set", na=False)
]
display(set_products[["Product", "Total_Sales"]])
```

## 6. Sorting

```python
# Sort by Total_Sales in ascending order
ascending_sales = df.sort_values("Total_Sales")
display(ascending_sales[["Order_ID", "Product", "Total_Sales"]].head(10))

# Sort by Total_Sales in descending order
descending_sales = df.sort_values("Total_Sales", ascending=False)
display(descending_sales[["Order_ID", "Product", "Total_Sales"]].head(10))

# Sort by Category and then Total_Sales
multi_sort = df.sort_values(
    ["Category", "Total_Sales"], ascending=[True, False]
)
display(multi_sort[["Category", "Product", "Total_Sales"]].head(15))
```

## 7. Basic Sales Statistics

```python
total_sales = df["Total_Sales"].sum()
average_sales = df["Total_Sales"].mean()
highest_sales = df["Total_Sales"].max()
lowest_sales = df["Total_Sales"].min()
total_quantity = df["Quantity"].sum()
number_of_orders = df["Order_ID"].nunique()

print(f"Total Sales: ₹{total_sales:,.2f}")
print(f"Average Sales per Order: ₹{average_sales:,.2f}")
print(f"Highest Order Sales: ₹{highest_sales:,.2f}")
print(f"Lowest Order Sales: ₹{lowest_sales:,.2f}")
print(f"Total Quantity Sold: {total_quantity}")
print(f"Number of Orders: {number_of_orders}")
```

## 8. Groupby and Aggregation – Category Analysis

```python
category_analysis = (
    df.groupby("Category")
      .agg(
          Orders=("Order_ID", "count"),
          Total_Quantity=("Quantity", "sum"),
          Total_Sales=("Total_Sales", "sum"),
          Average_Sales=("Total_Sales", "mean")
      )
      .sort_values("Total_Sales", ascending=False)
)
display(category_analysis)
```

## 9. Sales by City

```python
city_analysis = (
    df.groupby("City")
      .agg(
          Orders=("Order_ID", "count"),
          Total_Quantity=("Quantity", "sum"),
          Total_Sales=("Total_Sales", "sum"),
          Average_Sales=("Total_Sales", "mean")
      )
      .sort_values("Total_Sales", ascending=False)
)
display(city_analysis)
```

## 10. Sales by Product

```python
product_analysis = (
    df.groupby("Product")
      .agg(
          Orders=("Order_ID", "count"),
          Total_Quantity=("Quantity", "sum"),
          Total_Sales=("Total_Sales", "sum"),
          Average_Sales=("Total_Sales", "mean")
      )
      .sort_values("Total_Sales", ascending=False)
)
display(product_analysis)
```

## 11. Sales by Payment Method

```python
payment_analysis = (
    df.groupby("Payment_Method")
      .agg(
          Orders=("Order_ID", "count"),
          Total_Quantity=("Quantity", "sum"),
          Total_Sales=("Total_Sales", "sum"),
          Average_Sales=("Total_Sales", "mean")
      )
      .sort_values("Total_Sales", ascending=False)
)
display(payment_analysis)
```

## 12. Top-Performing Orders and Products

```python
print("Top 10 orders by sales:")
display(
    df.nlargest(10, "Total_Sales")[
        ["Order_ID", "Customer", "Product", "Category", "Total_Sales"]
    ]
)

print("Top 10 products by total sales:")
top_products = (
    df.groupby("Product")["Total_Sales"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
)
display(top_products)
```

## 13. Highest and Lowest Sales Orders

```python
highest_order = df.loc[df["Total_Sales"].idxmax()]
lowest_order = df.loc[df["Total_Sales"].idxmin()]

print("Highest sales order:")
display(highest_order.to_frame("Value"))

print("Lowest sales order:")
display(lowest_order.to_frame("Value"))
```

## 14. Meaningful Observations

```python
top_category = category_analysis["Total_Sales"].idxmax()
top_city = city_analysis["Total_Sales"].idxmax()
top_product = product_analysis["Total_Sales"].idxmax()
top_payment = payment_analysis["Total_Sales"].idxmax()

print(f"1. {top_category} generated the highest total sales among categories.")
print(f"2. {top_city} generated the highest total sales among cities.")
print(f"3. {top_product} was the top product by total sales.")
print(f"4. {top_payment} generated the highest total sales among payment methods.")
print(f"5. The dataset contains {number_of_orders} orders with total sales of ₹{total_sales:,.2f}.")
print(f"6. The average sales value per order is ₹{average_sales:,.2f}.")
```

## 15. Conclusion

The e-commerce dataset was analyzed using Pandas. Data selection, filtering, sorting, grouping, and aggregation were applied to identify sales patterns across categories, cities, products, and payment methods. The analysis also calculated total sales, average sales, highest and lowest order values, total quantity sold, and the number of orders.
